## 8. Cross-Sector Analysis

```python
import pandas as pd

summary_rows = []
for sector in ["agriculture", "education", "finance", "healthcare", "manufacturing", "retail"]:
    ndf = exp.node_df(sector)
    edf = exp.edge_df(sector)
    bdf = exp.belief_df(sector)
    summary_rows.append({
        "Sector": sector.title(),
        "Nodes": len(ndf),
        "Edges": len(edf),
        "Beliefs": len(bdf),
        "Obs Nodes": int(ndf["is_observation"].sum()),
        "Node Types": len(ndf["type"].unique()),
        "Edge Types": len(edf["type"].unique()),
        "Avg Belief Conf": round(bdf["confidence"].mean(), 3) if bdf["confidence"].notna().any() else None,
    })
pd.DataFrame(summary_rows).set_index("Sector")
```

```python
import matplotlib.pyplot as plt

sdf = pd.DataFrame(summary_rows).set_index("Sector")
fig, axes = plt.subplots(1, 3, figsize=(16, 5))

sdf["Nodes"].plot(kind="bar", ax=axes[0], color="#4A90D9", title="Node Count")
sdf["Edges"].plot(kind="bar", ax=axes[1], color="#E8743B", title="Edge Count")
sdf["Beliefs"].plot(kind="bar", ax=axes[2], color="#6AB04C", title="Belief Count")

for ax in axes:
    ax.tick_params(axis="x", rotation=45)
    ax.set_ylabel("Count")

plt.suptitle("MiNYAMA Sector Graph Comparison", fontsize=14, fontweight="bold")
plt.tight_layout()
plt.show()
```
